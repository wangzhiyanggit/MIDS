#### <span style="color:chocolate">Exercise 8:</span> Joining arrays (10 points)

Provide 8 examples that use the methods <span style="color:cornflowerBlue">np.concatenate()</span> and <span style="color:cornflowerBlue">np.stack()</span>. Note: create 4 examples using each method.

```python
a1 = np.array([1,2,3])
b1 = np.array([4,5,6])
concatenate_1 = np.concatenate((a1,b1))
print("Concatenate example 1:", concatenate_1)
print("shape:", concatenate_1.shape)

a2 = np.array([[1,2],[3,4]])
b2 = np.array([[5,6],[7,8]])
concatenate_2 = np.concatenate((a2,b2), axis = 0)
print("Concatenate example 2:", concatenate_2)
print("shape:", concatenate_2.shape)
concatenate_3 = np.concatenate((a2,b2), axis = 1)
print("Concatenate example 3:", concatenate_3)
print("shape:", concatenate_3.shape)

a3 = np.ones((2,2,1))
b3 = np.zeros((2,2,2))
concatenate_4 = np.concatenate((a3,b3), axis = 2)
print("Concatenate example 4:", concatenate_4)
print("shape:", concatenate_4.shape)

stack_1 = np.stack((a1,b1), axis = 0)
print("Stack example 1:", stack_1)
print("shape:", stack_1.shape)

stack_2 = np.stack((a1,b1), axis = 1)
print("Stack example 2:", stack_2)
print("shape:", stack_2.shape)

stack_3 = np.stack((a2,b2), axis = 0)
print("Stack example 3:", stack_3)
print("shape:", stack_3.shape)

a4 = np.ones((2,2)) * 255
b4 = np.ones((2,2)) * 120
c4 = np.ones((2,2)) * 0
stack_4 = np.stack((a4,b4,c4), axis = -1)
print("Stack example 4:", stack_4)
print("shape:", stack_4.shape)
```

#### <span style="color:chocolate">Numpy broadcasting</span> 

The term broadcasting describes how NumPy treats arrays during arithmetic operations. You can find more here:  https://numpy.org/doc/stable/user/basics.broadcasting.html

```python
# multiplying an array with a scalar
years = np.arange(1991, 2018, 1)
years * 1
```

```python
# adding a scalar
years + 2
```

```python
# many arithmetic operations are applied element wise on a pair of arrays
a = np.arange(3)
print('Array a:', a)
b = np.arange(3)
print('Array b:', b)
print('Sum of a and b:', a+b)
```

```python
# another example
a = np.ones((3, 3))
print('Array a:', a)
b = np.arange(3)
print('\nArray b:', b)
print('\nSum of a and b:', a+b)
```

#### <span style="color:chocolate">Exercise 9:</span> Numpy array standardization (10 points)

Consider the following (10, 4) array:

```python
my_array = np.random.random((10, 4))
print('my_array shape: ', my_array.shape)
my_array
```

Peform the following operations on this array:
1. Compute and print the mean and std for each column (note: set keepdims=True for proper broadcasting).
2. Standardize and print the array and its shape using the mean and std you computed above.

```python
column_means = my_array.mean(axis=0, keepdims=True)
column_stds = my_array.std(axis=0, keepdims=True)
print(f"1.Column means: {column_means}\n shape: {column_means.shape}")
print(f"Column stds: {column_stds}\n shape: {column_stds.shape}")

standardized_array = (my_array - column_means)/column_stds
print(f"2.Standardized array: {standardized_array}\n shape: {standardized_array.shape}")
```
